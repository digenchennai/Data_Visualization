```python
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
z_scores = zscore(data)
#outliers = [i for i in data if zscore(data) > 2]
outliers = [i for i, z in zip(data,z_scores) if abs(z) > 2]
outliers
```

```python
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
median = np.median(data)
z_scores = zscore(data)
#data = [median if abs(zscore(x)) > 2 else x for x in data]
cleaned_data = [median if abs(z) > 2 else x for x,z in zip(data,z_scores)]
print("Original data:", data)
print("Cleaned data:", cleaned_data)
```
