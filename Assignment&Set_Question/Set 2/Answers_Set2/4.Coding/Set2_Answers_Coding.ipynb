```python
'''
Question 1 :
An airline tracks flight delays (in minutes) for 20 flights. Analyze the flight delays to calculate percentiles, detect outliers, and evaluate the overall distribution.
'''
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

delays = [15, 30, 45, 20, 25, 100, 5, 60, 35, 50, 
          120, 80, 10, 75, 90, 200, 55, 40, 70, 150]

percentiles = np.percentile(delays,[10,25,50,75,90])
print("Percentiles (10th, 25th, 50th, 75th, 90th):",percentiles )
IQR = percentiles[3] - percentiles[1]
print("IQR:",IQR)
Outliers =  [x for x in delays if x < (percentiles[1] - 1.5 * IQR) or x > (percentiles[3] + 1.5 * IQR)]
print("Outliers:",Outliers)

plt.boxplot(delays, vert=False,
            patch_artist=True,
            boxprops=dict(facecolor='lightblue', color='black'),
            medianprops=dict(color='black', linewidth=2))
plt.title("Box Plot for Flight Delays")
plt.xlabel("Flight Delays (minutes)")
plt.show()
```

```python
'''
Question 2:

A company wants to analyze the salary distribution of its employees to understand the central tendency and determine whether the data is skewed.
DataSet:
'''
import numpy as np
from scipy import stats
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

salaries = [30000, 32000, 35000, 37000, 40000, 42000, 45000, 47000, 50000, 55000,
            60000, 62000, 65000, 67000, 70000, 72000, 75000, 80000, 85000, 90000]

mean_sal = np.mean(salaries)
median_sal = np.median(salaries)
mode_sal = stats.mode(salaries)
print("Mean Salary:",mean_sal, "\nMedian Salary:",median_sal, "\nMode Salary:",mode_sal[0])
print("The data is Left Skewed (Negative Skew)" if mean_sal < median_sal else "The data is Right Skewed (positive Skew)")

sns.histplot(salaries, bins=10, kde=True, color='skyblue', edgecolor='black')
plt.title("Histogram of Employee Salaries")
plt.xlabel("Salary")
plt.ylabel("Count")

```

```python
'''
Question 3:

 A school wants to analyze the exam performance of students across three subjects: Mathematics, Science, and English. 
 How can Data Science concepts be applied to understand their performance?
'''
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Mathematics': [85, 78, 92, 88, 70, 95, 60, 80, 90, 76],
    'Science': [80, 85, 88, 70, 75, 92, 55, 82, 89, 78],
    'English': [78, 74, 85, 80, 68, 90, 50, 77, 83, 72]
}

df = pd.DataFrame(data)

#.Descriptive Statistics
print(df[["Mathematics","Science","English"]].describe())


#. Histogram(graph)
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.hist(df['Mathematics'], bins=5, color='skyblue', edgecolor='black')
plt.title("Mathematics Score Distribution")

plt.subplot(1,3,2)
plt.hist(df['Science'],bins=5, color='lightgreen', edgecolor='black')
plt.title("Science Score Distribution")

plt.subplot(1,3,3)
plt.hist(df['English'], bins=5, color='salmon', edgecolor='black')
plt.title("English Score Distribution")

plt.tight_layout()
plt.show()

#. Correlation Analysis(graph)
correlation_matrix = df[['Mathematics','Science','English']].corr()
print("Correlation Analysis:\n",correlation_matrix)

sns.pairplot(df[['Mathematics','Science','English']], diag_kind="kde")
plt.suptitle("Correlation Analysis Scatterplots", y=1.02)
plt.show()

#. HeatMap(graph)

plt.figure(figsize=(6,4))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap of Subjects")
plt.show()
```

```python
'''
Question 4:
A pharmaceutical company conducted a clinical trial with two groups: one receiving medication and the other a placebo. 
How do you perform a hypothesis test to determine the effectiveness of the medication?
'''
from scipy.stats import ttest_ind
medication_group = [110, 115, 108, 102, 107, 99, 111, 104, 109, 101]
placebo_group = [120, 125, 130, 122, 128, 119, 124, 127, 123, 126]

t_stat, p_val = ttest_ind(medication_group,placebo_group)
print("T-Statistic:",t_stat)
print("P-Value:",p_val)
alpha = 0.05

print("Reject the null hypothesis: The medication is effective." if p_val < alpha else 
      "Fail to reject the null hypothesis, No strong evidence the medication is effective.")
```

```python
'''
Question 5 : A company conducted a customer satisfaction survey where customers rated their experience on a scale of 1 to 10. 
Analyze the survey results to calculate descriptive statistics and visualize the distribution of customer satisfaction ratings.
'''
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

ratings = [8, 9, 7, 5, 6, 10, 9, 4, 7, 8, 
           6, 9, 10, 5, 8, 7, 6, 9, 10, 7]

mean_rating = np.mean(ratings)
median_rating = np.median(ratings)
mode_rating = stats.mode(ratings)[0]
std_dev_rating = np.std(ratings)

print("Mean Rating:",mean_rating,"\nMedian Rating:",median_rating,"\nMode Rating:",mode_rating,"\nStandard Deviation:",std_dev_rating)

sns.histplot(ratings, bins=6, stat="frequency", kde=True, color='skyblue', edgecolor='black')
plt.title("Distribution of Customers Satisfaction Ratings")
plt.xlabel("Customer Satisfaction Rating")
plt.ylabel("Frequency")
plt.show()
```
