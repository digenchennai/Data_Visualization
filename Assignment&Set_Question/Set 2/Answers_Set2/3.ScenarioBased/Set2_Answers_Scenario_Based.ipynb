```python
'''
Scenario 1: 
Flight Delay Analysis
An airline tracks flight delays (in minutes) for 20 flights. 
Analyze the flight delays to calculate percentiles, detect outliers, and evaluate the overall distribution.
'''

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Example dataset: delays in minutes for 20 flights
delays = [5, 10, 12, 8, 15, 20, 25, 30, 35, 40,
          45, 50, 55, 60, 65, 70, 75, 80, 200, 250]

percentiles =  np.percentile(delays,[25,50,75])
print("25th percentile (Q1):",percentiles[0])
print("25th percentile (Median):",percentiles[1])
print("25th percentile (Q3):",percentiles[2]) 

Q1, Q3 = percentiles[0], percentiles[2]
IQR = Q3 - Q1
lower_band = Q1 - 1.5 * IQR
upper_band = Q3 + 1.5 * IQR

outliers = [x for x in delays if x < lower_band or x > upper_band]
print("Outliers:", outliers)

plt.boxplot(delays)
plt.title("Boxplot of Flight Delays")
plt.ylabel("Delay (minutes)")
```

```python
'''
Scenario 2: 
Employee Salary Analysis
A company wants to analyze the salary distribution of its employees to understand the central tendency and determine whether the data is skewed.
'''

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew
from scipy import stats

# Example dataset: salaries of 15 employees (in $)
salaries = [30000, 32000, 35000, 37000, 40000, 42000, 45000, 
            47000, 50000, 52000, 55000, 60000, 65000, 120000, 150000]

mean_salary = np.mean(salaries)
median_salary = np.median(salaries)
mode_salary = stats.mode(salaries)[0]

salary_skewness = skew(salaries)
print("Skewness:", salary_skewness)

plt.hist(salaries, bins=10, color="skyblue", edgecolor="black")
plt.title("Histogram of Employee Salaries")
plt.xlabel("Salary")
plt.ylabel("Frequency")
```

```python
'''
Scenario 3: 
Product Sales Analysis
A retail store records product sales over 15 days. Create a frequency distribution table and visualize the sales data using appropriate charts.
'''

import pandas as pd
import matplotlib.pyplot as plt

sales = [20, 25, 22, 30, 28, 35, 40, 22, 25, 30, 28, 35, 20, 40, 50]
freqTable = pd.DataFrame(columns = ["Unique_Values","Frequency","Relative_Frequency","Cusum"])

freqTable["Unique_Values"] = pd.Series(sales).value_counts().index
freqTable["Frequency"] = pd.Series(sales).value_counts().values
freqTable["Relative_Frequency"] = (freqTable["Frequency"]/15)
freqTable["Cusum"] = freqTable["Relative_Frequency"].cumsum()
print(freqTable)

plt.plot(freqTable["Unique_Values"], freqTable["Frequency"], color="blue", marker="o", label="Frequency")
plt.title("Sales Data Analysis")
plt.xlabel("Unique Values")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
'''
Scenario 4: 
Student Exam Performance Analysis
A school wants to analyze the exam performance of students across three subjects: Mathematics, Science, and English. 
How can Data Science concepts be applied to understand their performance?
'''

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Student": ["A","B","C","D","E","F","G","H"],
    "Math":    [85, 70, 95, 60, 75, 88, 92, 55],
    "Science": [78, 65, 88, 58, 72, 80, 90, 50],
    "English": [90, 75, 92, 65, 70, 85, 95, 60]
}

df = pd.DataFrame(data)
print("Mean scores:\n", df[["Math","Science","English"]].mean())
print("\nMedian scores:\n", df[["Math","Science","English"]].median())
print("\nStandard deviation::\n", df[["Math","Science","English"]].std())

sns.boxplot(data=df[["Math","Science","English"]])
plt.title("Score Distribution by Subject")
plt.show()
```

```python
'''
Scenario 5: 
Clinical Trial for Diabetes Medication
A pharmaceutical company conducted a clinical trial with two groups: one receiving medication and the other a placebo. Perform a hypothesis test to determine the effectiveness of the medication.
'''

import numpy as np
from scipy.stats import ttest_ind

medication_group = [25, 30, 28, 35, 40, 32, 31, 29, 34, 36]
placebo_group    = [10, 12, 15, 14, 11, 13, 9, 16, 12, 14]
alpha = 0.05
t_stat, p_value = ttest_ind(medication_group,placebo_group )
print("t_stat: ",t_stat)
print("p_value: ",p_value)

H0 = "Reject H0 → Medication is effective" if p_value < alpha else "Fail to reject H0 → No strong evidence medication works"
print(H0)
```
